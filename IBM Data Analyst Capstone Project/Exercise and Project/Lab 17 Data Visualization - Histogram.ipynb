## Hands-on Lab: Visualizing Data with Histograms


### 1. Visualizing the distribution of data (Histograms)


**1.1 Histogram of `CompTotal` (Total Compensation)**


Objective: Plot a histogram of `CompTotal` to visualize the distribution of respondents' total compensation.


```python
## Write your code here
QUERY = """
SELECT 
    CASE 
         WHEN CompTotal < 20000 THEN '0-20k'
         WHEN CompTotal >= 20000 AND CompTotal < 40000 THEN '20k-40k'
         WHEN CompTotal >= 40000 AND CompTotal < 60000 THEN '40k-60k'
         WHEN CompTotal >= 60000 AND CompTotal < 80000 THEN '60k-80k'
         WHEN CompTotal >= 80000 AND CompTotal < 100000 THEN '80k-100K'
         WHEN CompTotal >= 100000 AND CompTotal < 200000 THEN '100k-200k'
         WHEN CompTotal >= 200000 AND CompTotal < 500000 THEN '200k-500k'
         WHEN CompTotal >= 500000 AND CompTotal < 1000000 THEN '500k-1M'
         ELSE '1M+'
    END AS CompTotalRange,
    COUNT(*) AS Respondents
FROM main
WHERE CompTotal IS NOT NULL
GROUP BY CompTotalRange;
"""
df_CompTotal = pd.read_sql_query(QUERY, conn)
comp_order = ['0-20k', '20k-40k', '40k-60k', '60k-80k', '80k-100K', 
              '100k-200k', '200k-500k', '500k-1M', '1M+']
df_CompTotal['CompTotalRange'] = pd.Categorical(df_CompTotal['CompTotalRange'], categories=comp_order, ordered=True)
df_CompTotal = df_CompTotal.sort_values('CompTotalRange')
plt.figure(figsize=(10, 6))
plt.bar(df_CompTotal['CompTotalRange'], df_CompTotal['Respondents'], color='gold', edgecolor='black')
plt.xlabel('Total Compensation Range')
plt.ylabel('Respondents')
plt.title('Distribution of Total Compensation (Ordered by Range)')
plt.show()
```

**1.2 Histogram of YearsCodePro (Years of Professional Coding Experience)**


Objective: Plot a histogram of `YearsCodePro` to analyze the distribution of coding experience among respondents.


```python
## Write your code here
query = "SELECT YearsCodePro FROM main;"
df_Years_Professional_Code = pd.read_sql_query(query, conn)
plt.figure(figsize=(10, 6)) 
plt.hist(df_Years_Professional_Code['YearsCodePro'].dropna(), bins=20, color='gold', edgecolor='black') 
plt.title('Distribution of Professional Coding Experience')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

### 2. Visualizing Relationships in Data


**2.1 Histogram Comparison of `CompTotal` by `Age` Group**


Objective: Use histograms to compare the distribution of CompTotal across different Age groups.


```python
## Write your code here
query = """
SELECT 
    CompTotal,
    CASE 
        WHEN Age < 18 THEN '1. Under 18'
        WHEN Age BETWEEN 18 AND 34 THEN '2. 18-34'
        WHEN Age BETWEEN 35 AND 54 THEN '3. 35-54'
        ELSE '4. 55+' 
    END AS AgeGroup
FROM main
WHERE CompTotal > 0 AND CompTotal <= 500000
"""
df = pd.read_sql_query(query, conn)
plt.figure(figsize=(12, 6))
sns.histplot(data=df, x='CompTotal', hue='AgeGroup', element='step', kde=True)
plt.title('CompTotal Distribution')
plt.xlabel('Total Compensation')
plt.show()
```

**2.2 Histogram of TimeSearching for Different Age Groups**


Objective: Use histograms to explore the distribution of `TimeSearching` (time spent searching for information) for respondents across different age groups.


```python
## Write your code here
query = """
SELECT 
    CASE 
        WHEN Age < 18 THEN '1. Under 18'
        WHEN Age BETWEEN 18 AND 34 THEN '2. 18-34'
        WHEN Age BETWEEN 35 AND 54 THEN '3. 35-54'
        ELSE '4. 55+' 
    END AS AgeGroup,
    AVG(TimeSearching) as AvgTime
FROM main
GROUP BY AgeGroup
ORDER BY AgeGroup;
"""
df = pd.read_sql_query(query, conn)
plt.bar(df['AgeGroup'], df['AvgTime'], color='gold', edgecolor='black')
plt.title('Average Time Searching by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Time (minutes)')
plt.show()
```

### 3. Visualizing the Composition of Data


**3.1 Histogram of Most Desired Databases (`DatabaseWantToWorkWith`)**


Objective: Visualize the most desired databases for future learning using a histogram of the top 5 databases.


```python
## Write your code here
query = "SELECT DatabaseWantToWorkWith FROM main WHERE DatabaseWantToWorkWith IS NOT NULL"
df = pd.read_sql_query(query, conn)
database_counts = df['DatabaseWantToWorkWith'].str.split(';').explode().value_counts()
plt.figure(figsize=(12, 6))
database_counts.head(15).plot(kind='bar', color='gold')
plt.title('Most Desired Databases')
plt.xlabel('Database Technology')
plt.ylabel('Number of Respondents')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**3.2 Histogram of Preferred Work Locations (`RemoteWork`)**


Objective: Use a histogram to explore the distribution of preferred work arrangements (`remote work`).


```python
## Write your code here
query = "SELECT RemoteWork FROM main;"
df = pd.read_sql_query(query, conn)
df['RemoteWork'].hist(bins=5, color= 'blue', edgecolor='black')
plt.title('Preferred Remote Work Days Distribution')
plt.xlabel('Days Remote')
plt.ylabel('Frequency')
plt.show()
```

### 4. Visualizing Comparison of Data


**4.1 Histogram of Median CompTotal for Ages 45 to 60**


Objective: Plot the histogram for `CompTotal` within the age group 45 to 60 to analyze compensation distribution among mid-career respondents.


```python
## Write your code here
query = """
SELECT 
    CompTotal,
    CASE 
        WHEN Age < 18 THEN '1. Under 18'
        WHEN Age BETWEEN 18 AND 34 THEN '2. 18-34'
        WHEN Age BETWEEN 35 AND 54 THEN '3. 35-54'
        ELSE '4. 55+' 
    END AS AgeGroup
FROM main
WHERE CompTotal > 0 AND CompTotal <= 500000
"""
df = pd.read_sql_query(query, conn)
mid_career = df[df['AgeGroup'].isin(['3. 35-54', '4. 55+'])]
plt.figure(figsize=(10, 6))
plt.hist(mid_career['CompTotal'], bins=35, color='#4a90e2', edgecolor='white')
plt.title('Compensation Distribution (Ages 35-55+)')
plt.xlabel('Total Compensation (Capped at 500k)')
plt.ylabel('Number of Respondents')
plt.show()
```

**4.2 Histogram of Job Satisfaction (`JobSat`) by YearsCodePro**


Objective: Plot the histogram for `JobSat` scores based on respondents' years of professional coding experience.


```python
## Write your code here
query = """
SELECT JobSat, YearsCodePro 
FROM main
WHERE JobSat IS NOT NULL AND YearsCodePro IS NOT NULL
"""
df = pd.read_sql_query(query, conn)
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df = df.dropna()
plt.figure(figsize=(12, 6))
sns.histplot(data=df, x='YearsCodePro', hue='JobSat', multiple='stack', bins=30, palette='viridis')
plt.title('Distribution of Job Satisfaction by Professional Coding Experience')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Number of Respondents')
plt.show()
```
