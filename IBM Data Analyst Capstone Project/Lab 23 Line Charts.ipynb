```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Load the data


```python
df = pd.read_csv("survey-data.csv")
```

```python
df.head()
```

### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
## Write your code here
median_by_age = df.groupby("Age")["ConvertedCompYearly"].median()
plt.figure(figsize=(10,6))
plt.plot(median_by_age.index, median_by_age.values, marker='o', color='blue')
plt.title("Median Yearly Compensation by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Median ConvertedCompYearly")
plt.show()
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
age_groups = ["25-34 years old", "35-44 years old"]
filtered = df[df["Age"].isin(age_groups)].copy()
median_comp = (
    filtered.groupby("Age")["ConvertedCompYearly"]
    .median()
    .reindex(age_groups)
    .reset_index()
)
median_comp["AgeNumeric"] = [30, 40]
plt.figure(figsize=(10, 6))
sns.lineplot(
    data=median_comp,
    x="AgeNumeric",
    y="ConvertedCompYearly",
    marker="o",
    color='blue'
)
plt.title("Median ConvertedCompYearly for Ages 25 to 45")
plt.xlabel("Age")
plt.ylabel("Median ConvertedCompYearly")
plt.show()
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
df['WorkExp'] = pd.to_numeric(df['WorkExp'], errors='coerce')
df['JobSatPoints_6'] = pd.to_numeric(df['JobSatPoints_6'], errors='coerce')
exp_sat = df.groupby('WorkExp')['JobSatPoints_6'].median().dropna().reset_index()
plt.figure(figsize=(10,6))
sns.lineplot(data=exp_sat, x='WorkExp', y='JobSatPoints_6', marker="o", color='blue')
plt.title("Median Job Satisfaction by Experience Level")
plt.xlabel("Work Experience")
plt.ylabel("Median Job Satisfaction")
plt.show()
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
df['ConvertedCompYearly'] = pd.to_numeric(df['ConvertedCompYearly'], errors='coerce')
comp_exp = df.groupby('WorkExp')['ConvertedCompYearly'].median().dropna().reset_index()
plt.figure(figsize=(10,6))
sns.lineplot(data=comp_exp, x='WorkExp', y='ConvertedCompYearly',marker="o", color='blue')
plt.title("Median Compensation by Experience Level")
plt.xlabel("Work Experience (Years)")
plt.ylabel("Median Compensation (Yearly)")
plt.show()
```

##### 2.Line Chart of Job Satisfaction (`JobSatPoints_6`) Across Experience Levels

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.

- This chart will provide insight into how satisfaction correlates with experience over time


```python
sat_exp = df.groupby('WorkExp')['JobSatPoints_6'].median().dropna().reset_index()
plt.figure(figsize=(10,6))
sns.lineplot(data=sat_exp, x='WorkExp', y='JobSatPoints_6', marker="o", color='blue')
plt.title("Job Satisfaction Trends by Experience Level")
plt.xlabel("Work Experience (Years)")
plt.ylabel("Median Job Satisfaction")
plt.show()
```
