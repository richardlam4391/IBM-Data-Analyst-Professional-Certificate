**Read the CSV File into a Pandas DataFrame**

Load the Stack Overflow survey data into a Pandas DataFrame:


```python
# Read the CSV file
df = pd.read_csv('survey-data.csv')

# Display the first few rows of the data
df.head()
```

**Create a SQLite Database and Insert the Data**

Now, let's create a new SQLite database (`survey-data.sqlite`) and insert the data from the DataFrame into a table using the sqlite3 library:


```python
import sqlite3

# Create a connection to the SQLite database
conn = sqlite3.connect('survey-data.sqlite')

# Write the dataframe to the SQLite database
df.to_sql('main', conn, if_exists='replace', index=False)


# Close the connection
conn.close()
```

**Verify the Data in the SQLite Database**
Verify that the data has been correctly inserted into the SQLite database by running a simple query:


```python
# Reconnect to the SQLite database
conn = sqlite3.connect('survey-data.sqlite')

# Run a simple query to check the data
QUERY = "SELECT * FROM main LIMIT 5"
df_check = pd.read_sql_query(QUERY, conn)

# Display the results
print(df_check)
```

## Hands-on Lab


### Visualizing the Distribution of Data

**Histograms**

Plot a histogram of CompTotal (Total Compensation).


```python
QUERY1 = """
SELECT CompTotal 
FROM main 
WHERE CompTotal IS NOT NULL 
AND CompTotal > 0
AND CompTotal < 5000000  
"""
df1 = pd.read_sql_query(QUERY1, conn)
df1= df1[df1['CompTotal'] < 2000000]
plt.figure(figsize=(10, 6))
sns.histplot(data=df1, x='CompTotal', bins=20, color='blue')
plt.title('Histogram of Total Compensation')
plt.xlabel('Total Compensation')
plt.ylabel('Respondent Count')
plt.show()
```

**Box Plots**

Plot a box plot of Age.


```python
QUERY2 = """
SELECT Age 
FROM main
WHERE Age IS NOT NULL 
"""
df2 = pd.read_sql_query(QUERY2, conn)
plt.figure(figsize=(10, 6))
sns.boxplot(y=df2['Age'], color='blue')
plt.title('Distribution of Respondent Ages')
plt.ylabel('Age')
plt.show()
```

### Visualizing Relationships in Data

**Scatter Plots**

Create a scatter plot of Age and WorkExp.


```python
QUERY3 = """
SELECT Age, WorkExp 
FROM main 
WHERE Age IS NOT NULL 
AND WorkExp IS NOT NULL 
"""
df3 = pd.read_sql_query(QUERY3, conn)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df3, x='Age', y='WorkExp', alpha=0.6, color='blue')
plt.title('Scatter Plot: Age vs. Work Experience')
plt.xlabel('Age')
plt.ylabel('Work Experience (Years)')
plt.show()
```

**Bubble Plots**

Create a bubble plot of `TimeSearching` and `Frustration` using the Age column as the bubble size.


```python
QUERY4 = """
SELECT TimeSearching, Frustration, Age 
FROM main 
WHERE TimeSearching IS NOT NULL 
  AND Frustration IS NOT NULL 
  AND Age IS NOT NULL
"""
df4 = pd.read_sql_query(QUERY4, conn)
df4['Frustration'] = df4['Frustration'].str.split(';')
df4 = df4.explode('Frustration')
df4['Frustration'] = df4['Frustration'].str.strip()
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df4, 
    x='TimeSearching', 
    y='Frustration', 
    size='Age',
    hue='Age', 
    sizes=(20, 400),  
    palette='viridis',
    legend='brief'
)
plt.title('Bubble Plot: Time Searching vs. Frustration (Size = Age)')
plt.xlabel('Time Searching')
plt.ylabel('Frustration')
plt.show()
```

### Visualizing Composition of Data

**Pie Charts**

Create a pie chart of the top 5 databases(`DatabaseWantToWorkWith`) that respondents wish to learn next year.


```python
QUERY5 = """
SELECT DatabaseWantToWorkWith 
FROM main
WHERE DatabaseWantToWorkWith IS NOT NULL
"""
df5 = pd.read_sql_query(QUERY5, conn)
db_counts = {}
for tech_row in df5['DatabaseWantToWorkWith']:
    technologies = [t.strip() for t in tech_row.split(';') if t.strip()]
    for tech in technologies:
        db_counts[tech] = db_counts.get(tech, 0) + 1
sorted_dbs = sorted(db_counts.items(), key=lambda x: x[1], reverse=True)
top_5 = sorted_dbs[:5]
others_count = sum(count for name, count in sorted_dbs[5:])
labels = [name for name, count in top_5] + ['Others']
sizes = [count for name, count in top_5] + [others_count]
plt.figure(figsize=(10, 6))
plt.pie(
    sizes, 
    labels=labels, 
    autopct='%1.1f%%', 
    startangle=140, 
)
plt.title('Top 5 Databases Wanted for Next Year')
plt.show()
```

**Stacked Charts** 

Create a stacked bar chart of median `TimeSearching` and `TimeAnswering` for the age group 30 to 35.


```python
QUERY6 = """
SELECT TimeSearching, TimeAnswering
FROM main
WHERE Age = '25-34 years old'
AND TimeSearching IS NOT NULL
AND TimeAnswering IS NOT NULL
"""
df6 = pd.read_sql_query(QUERY6, conn)
midpoints = {
    'Less than 15 minutes a day': 7.5,
    '15-30 minutes a day':        22.5,
    '30-60 minutes a day':        45,
    '60-120 minutes a day':       90,
    'Over 120 minutes a day':     150,
}
df6['SearchMin'] = df6['TimeSearching'].map(midpoints)
df6['AnswerMin'] = df6['TimeAnswering'].map(midpoints)
median_search = df6['SearchMin'].median()
median_answer = df6['AnswerMin'].median()
plot_df = pd.DataFrame({
    'Age Group':      ['25-34 (includes 30–35)'],
    'TimeSearching':  [median_search],
    'TimeAnswering':  [median_answer],
})
colors = sns.color_palette('Set2')
fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(plot_df['Age Group'], plot_df['TimeSearching'],
       label='Median TimeSearching', color='blue')
ax.bar(plot_df['Age Group'], plot_df['TimeAnswering'],
       bottom=plot_df['TimeSearching'],
       label='Median TimeAnswering', color='gold')
ax.set_ylabel('Minutes per day')
ax.set_title('Median Time Searching & Answering — Age 30–35')
ax.legend()
for i, (s, a) in enumerate(zip(plot_df['TimeSearching'], plot_df['TimeAnswering'])):
    ax.text(i, s / 2,     f'{s:.0f}', ha='center', va='center',
            color='black')
    ax.text(i, s + a / 2, f'{a:.0f}', ha='center', va='center',
            color='black')
plt.show()
```

### Visualizing Comparison of Data

**Line Chart**

Plot the median `CompTotal` for all ages from 45 to 60.


```python
QUERY7 = """
SELECT Age, CompTotal 
FROM main
WHERE Age BETWEEN 45 AND 60 
  AND CompTotal IS NOT NULL
"""
df7 = pd.read_sql_query(QUERY7, conn)
median_df = df7.groupby('Age')['CompTotal'].median().reset_index()
plt.figure(figsize=(10, 6))
sns.lineplot(
    x='Age', 
    y='CompTotal', 
    data=median_df, 
    marker='o',
    markersize=8,
    linewidth=2.5,
    color='blue'
)
plt.title('Median Total Compensation for Ages 45 to 60')
plt.xlabel('Age Group')
plt.ylabel('Median CompTotal')
plt.show()
```

**Bar Chart**

Create a horizontal bar chart using the `MainBranch` column.


```python
QUERY8 = """
SELECT MainBranch, COUNT(*) as Count 
FROM main
GROUP BY MainBranch 
ORDER BY Count DESC
"""
df8 = pd.read_sql_query(QUERY8, conn)
plt.figure(figsize=(10, 6))
sns.barplot(data=df8, x='Count', y='MainBranch', hue='MainBranch', palette='viridis')
plt.title('Respondent Distributions')
plt.xlabel('Number of Respondents')
plt.ylabel('Main Branch')
plt.show()
```
