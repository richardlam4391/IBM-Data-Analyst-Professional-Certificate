### Step 1: Import Libraries and Load Data


- Import the `pandas`, `matplotlib.pyplot`, and `seaborn` libraries.

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

```python
job_counts = df["JobSat"].value_counts()
plt.figure(figsize=(10,6))
plt.pie(job_counts, labels=job_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Job Satisfaction Distribution")
plt.axis('equal')
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
have = df["LanguageHaveWorkedWith"].dropna().str.split(";")
want = df["LanguageWantToWorkWith"].dropna().str.split(";")
have_counts = Counter([lang.strip() for sublist in have for lang in sublist])
want_counts = Counter([lang.strip() for sublist in want for lang in sublist])
```

```python
plt.figure(figsize=(8,6))
venn2(
    subsets = (
        len(set(have_counts.keys()) - set(want_counts.keys())),
        len(set(want_counts.keys()) - set(have_counts.keys())),
        len(set(have_counts.keys()) & set(want_counts.keys()))
    ),
    set_labels = ("Have Worked With", "Want To Work With")
)
plt.title("Overlap of Programming Languages")
plt.show()
```

```python
lang_df = pd.DataFrame({
    "Language": list(set(have_counts.keys()) | set(want_counts.keys())),
    "HaveWorkedWith": [have_counts.get(lang, 0) for lang in set(have_counts.keys()) | set(want_counts.keys())],
    "WantToWorkWith": [want_counts.get(lang, 0) for lang in set(have_counts.keys()) | set(want_counts.keys())]
})
top10 = lang_df.sort_values("HaveWorkedWith", ascending=False).head(10)
plt.figure(figsize=(10,6))
plt.bar(top10["Language"], top10["HaveWorkedWith"], label="Have Worked With", color='blue')
plt.bar(top10["Language"], top10["WantToWorkWith"], label="Want To Work With", color='gold')
plt.title("Top 10 Programming Languages: Have Worked With vs Want To Work With")
plt.ylabel("Count")
plt.legend()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
top_10_countries = df['Country'].value_counts().nlargest(10).index
df_top10 = df[df['Country'].isin(top_10_countries)]
fig, ax = plt.subplots(figsize=(10, 6))
sns.countplot(
    data=df_top10, 
    x='Country', 
    hue='RemoteWork', 
    order=top_10_countries, 
    palette='viridis',
    ax=ax
)
ax.set_title('Distribution of Remote Work by Top 10 Regions')
ax.set_xlabel('Country')
ax.set_ylabel('Count')
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
df["JobSat"] = df["JobSat"].astype(str).str.strip().str.lower()
job_map = {
    "very dissatisfied": 1,
    "dissatisfied": 2,
    "neither satisfied nor dissatisfied": 3,
    "satisfied": 4,
    "very satisfied": 5
}
df["JobSat_num"] = df["JobSat"].map(job_map)
def convert_code_years(x):
    if pd.isna(x):
        return np.nan
    x = str(x).strip().lower()
    if "less than" in x:
        return 0.5
    if "more than" in x:
        return 55
    try:
        return float(x)
    except:
        return np.nan
df["YearsCodePro_num"] = df["YearsCodePro"].apply(convert_code_years)
clean_df = df[["JobSat_num", "YearsCodePro_num"]].dropna()
if len(clean_df) < 2:
    print("Not enough valid data to compute correlation.")
    print("Valid rows found:", len(clean_df))
    print(clean_df.head())
else:
    pearson_corr, pearson_p = pearsonr(clean_df["JobSat_num"], clean_df["YearsCodePro_num"])
    spearman_corr, spearman_p = spearmanr(clean_df["JobSat_num"], clean_df["YearsCodePro_num"])
    print("Pearson Correlation:", pearson_corr)
    print("Pearson p-value:", pearson_p)
    print("\nSpearman Correlation:", spearman_corr)
    print("Spearman p-value:", spearman_p)
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
df_exploded = df.copy()
df_exploded["Employment"] = df_exploded["Employment"].str.split(";")
df_exploded = df_exploded.explode("Employment")
df_exploded = df_exploded.reset_index(drop=True)
df_exploded["Employment"] = df_exploded["Employment"].str.strip()
crosstab = pd.crosstab(
    df_exploded["EdLevel"], df_exploded["Employment"], normalize="index"
)
print(crosstab)
crosstab.plot(kind="bar", stacked=True, figsize=(10,6), colormap="viridis")
plt.title("Education vs Employment")
plt.title("Top 10 Programming Languages: Have Worked With vs Want To Work With")
plt.xlabel("Education Level")
plt.ylabel("Count")
plt.show()
```
