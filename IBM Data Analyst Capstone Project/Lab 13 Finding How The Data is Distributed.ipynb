### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
## Write your code here
plt.figure(figsize=(8, 8))
plt.pie(
    Job_Satisfaction_count, 
    labels=Job_Satisfaction_count.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=sns.color_palette('Set2')
)
plt.title('Distribution of Job Satisfaction (JobSat)')
plt.show()
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
## Write your code here
Language_Have_Work_With_count=df['LanguageHaveWorkedWith'].value_counts()
Language_Have_Work_With_count
```

```python
Language_Want_To_Work_With_count=df['LanguageWantToWorkWith'].value_counts()
Language_Want_To_Work_With_count
```

```python
worked_with = df['LanguageHaveWorkedWith'].str.split(';').explode().value_counts()
want_to_work = df['LanguageWantToWorkWith'].str.split(';').explode().value_counts()
comparison_df = pd.DataFrame({
    'Worked With': worked_with,
    'Want to Work With': want_to_work
}).fillna(0)
top_languages = comparison_df.sort_values(by='Worked With', ascending=False).head(10)
ax = top_languages.plot(kind='bar', figsize=(12, 6))
ax.set_title('Top 10 Languages: Currently Using vs. Desired')
ax.set_xlabel('Programming Language')
ax.set_ylabel('Number of Respondents')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
## Write your code here
Remote_Work_count=df['RemoteWork'].value_counts()
Remote_Work_count
```

```python
Region_count=df['Country'].value_counts()
Region_count
```

```python
grouped_data = df.groupby(['Country', 'RemoteWork']).size().reset_index(name='Count')
plt.figure(figsize=(10, 6))
sns.barplot(data=grouped_data, x='Country', y='Count', hue='RemoteWork')
plt.title('Distribution of Remote Work by Region')
plt.ylabel('Number of Employees')
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
## Write your code here
df.select_dtypes(include=['number']).corr()
```

```python
from scipy.stats import pearsonr
```

```python
df['YearsCodePro'] = df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 51})
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')
df['JobSat'] = pd.to_numeric(df['JobSat'], errors='coerce')
temp_df = df.dropna(subset=['YearsCodePro', 'JobSat'])
years_code_pro = temp_df['YearsCodePro']
job_sat = temp_df['JobSat']
pearson_coef, p_value = pearsonr(years_code_pro, job_sat)
print(f"The Pearson Correlation Coefficient is {pearson_coef:.3f}, with a P-value of P = {p_value:.3f}")
```
